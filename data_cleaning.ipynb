```python
import numpy as np
import pandas as pd
```

```python
# reading in the data
df = pd.read_csv("police_use_of_force.csv")
print(df.shape)
df.head(10)
```

```python
# finding all the null values
# the important variables are ForceType, Sex, Race, TypeOfResistance, Neighborhood, Problem, PrimaryOffense
# maybe we also want TypeOfResistance? 
df.isnull().sum()
```

```python
# dropping rows with null values in important variables
important = ['ForceType', 'Sex', 'Race', 'TypeOfResistance', 'Neighborhood', 'Problem', 'PrimaryOffense']
df_cleaned = df.dropna(subset=important)
df_cleaned.shape
```

With dropping all of the rows with any nulls in the 'important' columns, we still have almost 30,000 rows of data. 

```python
# now getting a df that only has the needed feature variables and target ('ForceType)
# all of these are categorical variables also
df_final = df_cleaned[important]
df_final
```

```python
# saving the cleaned data as a new csv
df_final.to_csv("cleaned_data.csv")
```
