## Further Data Cleaning

We already dropped the values that had a null ForceType and null features, but now we need to clean up/group the categories we want to use for analysis to make everything more feasible. 

```python
# reading in the data and looking at our target variable
# it looks like we are going to want to group categories
df = pd.read_csv('cleaned_data.csv')
print(df['ForceType'].value_counts())
print(df.shape)

df['ForceType'] = df['ForceType'].replace('Less Lethal Projectile', 'Less Lethal Weapons')
df['ForceType'] = df['ForceType'].replace('Baton', 'Less Lethal Weapons')
df['ForceType'] = df['ForceType'].replace('Less Lethal', 'Less Lethal Weapons')
df['ForceType'] = df['ForceType'].replace('Taser', 'Less Lethal Weapons')
df['ForceType'] = df['ForceType'].replace('Improvised Weapon', 'Less Lethal Weapons')
df['ForceType'] = df['ForceType'].replace('Police K9 Bite', 'Less Lethal Weapons')
df['ForceType'] = df['ForceType'].replace('Gun Point Display', 'Guns')
df['ForceType'] = df['ForceType'].replace('Firearm', 'Guns')
df['ForceType'] = df['ForceType'].replace('Maximal Restraint Technique', 'Bodily Force')

print(df['ForceType'].value_counts())
```

less lethal projectile, baton, less lethal, taser, improvised weapon, police k9 bite  - 'less-lethal weapons'

gun point display and firearm together - 'guns'

add maximal restraint technique to bolidy force - 'bodily force'

chemical irritant by itself

```python
# now looking at race - we also will want to group these
df['Race'] = df['Race'].replace('Other / Mixed Race', 'other')
df['Race'] = df['Race'].replace('Pacific Islander', 'other')
df['Race'] = df['Race'].replace('Asian', 'other')
df['Race'] = df['Race'].replace('Unknown', 'other')
df = df[df['Race'] != 'not recorded']
print(df['Race'].value_counts())
```

Group other/mixed race, asian, unknown,  pacific islander together and drop 'not recorded'

```python
df = df[df['Sex'] != 'not recorded']
print(df['Sex'].value_counts())
```

```python
# TypeOfResistance was giving me problems, so I am looking at
# all of the unique values
df['TypeOfResistance'].unique()
```

This data ^ is super messy and will need to be cleaned before building models. 

```python
# stripping the responses down since they are not clean enough to see 
# first removing capitalization and spaces 
df['TypeOfResistance'] = df['TypeOfResistance'].str.strip().str.lower()

# mapping phrasing together 
df['TypeOfResistance'] = df['TypeOfResistance'].replace({
    'commission of a crime': 'commission of crime',
    'fled in a vehicle': 'fled in vehicle',
    'assaulted police horse': 'assaulting police horse',
    'assaulted police k9': 'assaulting police k9'
})

# final cleaner results
print(df['TypeOfResistance'].value_counts())
```

```python
df = df[df['TypeOfResistance'] != 'x']
df['TypeOfResistance'] = df['TypeOfResistance'].replace('assaulting police horse', 'assaulting police officer or animal')
df['TypeOfResistance'] = df['TypeOfResistance'].replace('assaulting police k9', 'assaulting police officer or animal')
df['TypeOfResistance'] = df['TypeOfResistance'].replace('assaulted officer', 'assaulting police officer or animal')
df['TypeOfResistance'] = df['TypeOfResistance'].replace('unspecified', 'other/unspecified')
df['TypeOfResistance'] = df['TypeOfResistance'].replace('other', 'other/unspecified')
df['TypeOfResistance'] = df['TypeOfResistance'].replace('fled on foot', 'fled')
df['TypeOfResistance'] = df['TypeOfResistance'].replace('fled in vehicle', 'fled')
df['TypeOfResistance'].value_counts()
```
